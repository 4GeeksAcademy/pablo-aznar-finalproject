```python
import pandas as pd
from glob import glob
import os
```

```python
data_dir = '../data/raw'
# Buscar todos los CSV de la carpeta
accident_files = sorted(glob(os.path.join(data_dir, "*.csv")))
accident_files
```

```python
# Leer y unir todos los archivos
df_accidents = pd.concat([pd.read_csv(file) for file in accident_files], ignore_index=True)
df_accidents
```

```python
# Ver si hay valores duplicados del numero de expediente
duplicados = df_accidents[df_accidents["Numero_expedient"].duplicated()]
duplicados
```

```python
# Encontrar los expedientes duplicados (que aparecen más de una vez)
duplicados_ids = df_accidents["Numero_expedient"][df_accidents["Numero_expedient"].duplicated(keep=False)]
duplicados_ids
```

```python
# Filtrar todos los registros con los expedientes duplicados
expedientes_duplicados= df_accidents[df_accidents["Numero_expedient"].isin(duplicados_ids)]
expedientes_duplicados.sort_values("Numero_expedient")
```

```python
df_accidents.duplicated().sum()
```

```python
# Actualiza la columna existente Any con los valores de NK_Any cuando Any está vacía (NaN)
df_accidents["Any"] = df_accidents["Any"].combine_first(df_accidents["NK_Any"])

# Eliminar la columna redundante
df_accidents = df_accidents.drop(columns=["NK_Any"])
```

```python
df_accidents.shape
```

```python
# Tenemos la variable Num_postal duplicada con un espacion al final
[col for col in df_accidents.columns if "Num_postal" in col]
```

```python
# Creamos columna unificada, priorizando la que ya se llama exactamente "Num_postal" que no tiene espacio
df_accidents["Num_postal"] = df_accidents["Num_postal"].combine_first(df_accidents["Num_postal "])
```

```python
# Eliminamos variable redundante mal nombrada con un espacio al final
df_accidents.drop(columns=["Num_postal "], inplace=True)
```

```python
# Filtrar filas donde ambas coordenadas WGS84 no son nulas
filtro_wgs84 = (
    df_accidents["Longitud_WGS84"].notna() &
    df_accidents["Latitud_WGS84"].notna()
)

# Contar registros por año con datos WGS84 completos
wgs84_por_año = df_accidents[filtro_wgs84].groupby("Any").size()

wgs84_por_año
```
